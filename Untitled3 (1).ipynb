```python
import json, pandas as pd, run_nsight as rn
from pathlib import Path
out, copy_gbps, rows = Path("results/v2/nsight"), 1390.0, []
for log in sorted(out.glob("*_ncu.csv")):
    name = log.name[:-len("_ncu.csv")]
    meta_p = out / f"{name}_meta.json"
    try:
        df = rn.parse_ncu_csv(log); meta = json.loads(meta_p.read_text())
    except Exception as e:
        print(f"skip {name}: {e}"); continue
    if df.empty: print(f"skip {name}: no kernels"); continue
    s = {"target": name, **rn.summarize_kernels(df), "wall_ms": meta["wall_ms"]}
    s["gpu_busy_fraction"] = s["kernel_time_ms"] / meta["wall_ms"]
    s["dram_gbps_over_wall_time"] = s["dram_total_mb"] / meta["wall_ms"]
    mb = meta.get("model_bytes_total") or meta.get("model_bytes_min")
    if mb: s["model_estimate_mb"] = mb / 1e6; s["measured_over_model"] = s["dram_total_mb"] * 1e6 / mb
    s["achievable_copy_gbps"] = copy_gbps
    s["pct_of_achievable_over_kernel_time"] = 100 * s["dram_gbps_over_kernel_time"] / copy_gbps
    rows.append(s)
    df.to_csv(out / f"{name}_kernels.csv", index=False)
S = pd.DataFrame(rows); S.to_csv(out / "summary.csv", index=False)
print(S[["target","kernels","dram_total_mb","model_estimate_mb","measured_over_model",
         "dram_gbps_over_kernel_time","pct_of_achievable_over_kernel_time","gpu_busy_fraction"]].round(3).to_string())

# where do the bytes go? top kernels by DRAM traffic
for t in ["decode_tinyllama_b1_ctx1024", "decode_tinyllama_b32_ctx1024", "prefill_tinyllama_b8_p512"]:
    k = pd.read_csv(out / f"{t}_kernels.csv")
    k["mb"] = (k["dram__bytes_read.sum"] + k["dram__bytes_write.sum"]) / 1e6
    k["name"] = k.kernel.str.slice(0, 70)
    g = k.groupby("name").agg(calls=("mb", "size"), total_mb=("mb", "sum"),
                              time_ms=("gpu__time_duration.sum", lambda x: x.sum() / 1e6)).sort_values("total_mb", ascending=False)
    print(f"\n=== {t}: top kernels by DRAM MB (total {k.mb.sum():.0f} MB)"); print(g.head(12).round(2).to_string())
```
